# Logit Lens: Watching Predictions Form in Gemma 4

**Paper:** nostalgebraist, "interpreting GPT: the logit lens" (2020)
**Paper:** Belrose et al., "Eliciting Latent Predictions from Transformers with the Tuned Lens" (2023)

## What Is the Logit Lens?

Normally, only the **final layer's** hidden state gets projected through the LM head to produce token predictions. But what if we applied that same projection to **every intermediate layer**?

The Logit Lens does exactly that -- it reveals how the model's prediction evolves from vague semantics in early layers to confident predictions in later layers. You can literally watch the model "make up its mind."

### Key Insight
In a residual-stream transformer, each layer **refines** the representation. By decoding at each layer, we see the prediction **crystallize** over depth.

### Gemma 4 E4B
Gemma 4 E4B has **42 transformer layers**, giving us a fine-grained view of how predictions form across a deep network. We use the `AutoProcessor` interface (which wraps a tokenizer) rather than a standalone tokenizer.

## Setup

```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
import numpy as np
import os
import time
from transformers import AutoModelForCausalLM, AutoProcessor

%matplotlib inline

plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": "#f8f9fa",
    "axes.grid": True,
    "grid.alpha": 0.3,
    "font.size": 11,
    "axes.titlesize": 14,
    "axes.titleweight": "bold",
})

OUTPUT_DIR = os.path.expanduser("~/hook_outputs/charts")
os.makedirs(OUTPUT_DIR, exist_ok=True)

MODEL_ID = "google/gemma-4-E4B-it"
CACHE_DIR = os.path.expanduser("~/models")

print("Setup complete.")
print("Model: {}".format(MODEL_ID))
print("Output dir: {}".format(OUTPUT_DIR))
```

## Load Model

Gemma 4 uses `AutoProcessor` which wraps the tokenizer. We load with `dtype="auto"` and `device_map="auto"` so the model picks the best dtype and places layers across available devices.

```python
print("Loading processor and model...")
t0 = time.time()

processor = AutoProcessor.from_pretrained(MODEL_ID, cache_dir=CACHE_DIR)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID, cache_dir=CACHE_DIR, dtype="auto", device_map="auto",
)
model.eval()

print("Loaded in {:.1f}s".format(time.time() - t0))
print("Model dtype: {}".format(next(model.parameters()).dtype))
print("Num layers: {}".format(model.config.text_config.num_hidden_layers))
```

## Step 1: Capture Hidden States at Every Layer

We attach a forward hook to each of the 42 transformer layers to capture the residual stream output. We also hook the embedding layer to get the "layer -1" representation. Then we apply the model's own final RMSNorm + LM head to decode each layer's hidden state into vocabulary logits.

```python
PROMPT = "The capital of France is"

inputs = processor(text=PROMPT, return_tensors="pt").to(model.device)
input_ids = inputs["input_ids"]
tokens = [processor.decode([tid]) for tid in input_ids[0].tolist()]
num_layers = model.config.text_config.num_hidden_layers

print("Prompt: '{}'".format(PROMPT))
print("Tokens: {}".format(tokens))
print("Layers: {}".format(num_layers))

# Capture hidden states at every layer using hooks
layer_hidden_states = {}

def make_hook(layer_idx):
    def hook_fn(module, input, output):
        hidden = output[0] if isinstance(output, tuple) else output
        hidden = hidden.detach()
        layer_hidden_states[layer_idx] = hidden
    return hook_fn

handles = []
for i in range(num_layers):
    h = model.model.language_model.layers[i].register_forward_hook(make_hook(i))
    handles.append(h)

# Also capture embedding output (layer -1)
def embed_hook(module, input, output):
    hidden = output.detach()
    layer_hidden_states[-1] = hidden
handles.append(model.model.language_model.embed_tokens.register_forward_hook(embed_hook))

with torch.no_grad():
    outputs = model(**inputs)

for h in handles:
    h.remove()

print("Captured hidden states from {} layers + embedding".format(len(layer_hidden_states)))
# Show shapes to verify
for key in [-1, 0, num_layers - 1]:
    if key in layer_hidden_states:
        print("  Layer {}: shape {}".format(key, layer_hidden_states[key].shape))
```

## Step 2: Apply the Logit Lens

For each layer, we take the hidden state, apply the final RMSNorm, then project through the LM head to get vocabulary logits. This tells us: **"If decoding stopped at this layer, what would the model predict?"**

We handle both 2D `(seq_len, hidden_dim)` and 3D `(batch, seq_len, hidden_dim)` tensor shapes since different model configurations may output either format.

```python
# Apply logit lens: final_norm + lm_head at each layer
logit_lens_results = {}  # layer_idx -> {token_pos -> (top_token, probability)}

final_norm = model.model.language_model.norm
lm_head = model.lm_head

with torch.no_grad():
    for layer_idx in sorted(layer_hidden_states.keys()):
        hidden = layer_hidden_states[layer_idx]
        # Handle 2D case: unsqueeze to (1, seq_len, hidden_dim)
        if hidden.dim() == 2:
            hidden = hidden.unsqueeze(0)
        # Apply final norm then LM head
        normed = final_norm(hidden)
        logits = lm_head(normed)
        # Handle 2D logits case
        if logits.dim() == 2:
            logits = logits.unsqueeze(0)
        probs = F.softmax(logits[0].float(), dim=-1)  # (seq_len, vocab_size)
        
        layer_results = {}
        for pos in range(probs.shape[0]):
            top_probs, top_ids = probs[pos].topk(5)
            top_tokens = [processor.decode([tid.item()]) for tid in top_ids]
            layer_results[pos] = list(zip(top_tokens, top_probs.cpu().tolist()))
        
        logit_lens_results[layer_idx] = layer_results

# Show the last token position (next-token prediction) across layers
print("Logit Lens: Next-token prediction at each layer")
print("=" * 70)
print("{:<10} {:<20} {:<10} {:<30}".format("Layer", "Top Prediction", "Prob", "Top-5"))
print("-" * 70)
for layer_idx in sorted(logit_lens_results.keys()):
    result = logit_lens_results[layer_idx]
    last_pos = max(result.keys())
    top1_token, top1_prob = result[last_pos][0]
    top5_str = ", ".join(["{} ({:.2f})".format(t.strip(), p) for t, p in result[last_pos][:3]])
    label = "Embed" if layer_idx == -1 else "L{}".format(layer_idx)
    print("{:<10} {:<20} {:<10.4f} {:<30}".format(label, repr(top1_token.strip()), top1_prob, top5_str))
```

## Step 3: Visualize -- The Prediction Heatmap

The signature logit lens visualization: a two-panel figure showing:
1. **P(correct next token)** at each (layer, position) pair
2. **Top-1 confidence** at each (layer, position) pair

Watch predictions crystallize from noise to confident answers across Gemma 4's 42 layers.

```python
# Build the probability matrix for the CORRECT next token at each (layer, position)
seq_len = len(tokens)

# Get the actual next tokens (shifted by 1)
actual_next_tokens = input_ids[0, 1:].cpu().tolist()

# Matrix: probability of the correct next token at each layer
correct_token_probs = np.zeros((num_layers, seq_len - 1))

# Matrix: top-1 predicted token text and confidence at each layer
top1_text = np.empty((num_layers, seq_len), dtype=object)
top1_prob_matrix = np.zeros((num_layers, seq_len))

with torch.no_grad():
    for li in range(num_layers):
        hidden = layer_hidden_states[li]
        if hidden.dim() == 2:
            hidden = hidden.unsqueeze(0)
        normed = final_norm(hidden)
        logits = lm_head(normed)
        if logits.dim() == 2:
            logits = logits.unsqueeze(0)
        probs = F.softmax(logits[0].float(), dim=-1)
        
        for pos in range(seq_len - 1):
            next_token_id = actual_next_tokens[pos]
            correct_token_probs[li, pos] = probs[pos, next_token_id].cpu().item()
        
        for pos in range(seq_len):
            top_prob, top_id = probs[pos].max(dim=-1)
            top1_text[li, pos] = processor.decode([top_id.item()]).strip()
            top1_prob_matrix[li, pos] = top_prob.cpu().item()

print("Built logit lens matrices: {} layers x {} positions".format(num_layers, seq_len))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(20, 12))
fig.suptitle("Logit Lens: Gemma 4 E4B Predictions", fontsize=16, fontweight="bold")

# Left: Probability of correct next token
ax = axes[0]
im = ax.imshow(correct_token_probs, aspect="auto", cmap="YlOrRd", interpolation="nearest", vmin=0, vmax=1)
ax.set_xlabel("Token Position")
ax.set_ylabel("Layer")
ax.set_title("P(correct next token) at each layer")
ax.set_xticks(range(seq_len - 1))
ax.set_xticklabels(
    ["{} -> {}".format(tokens[i].strip(), tokens[i+1].strip()) for i in range(seq_len - 1)],
    rotation=45, ha="right", fontsize=8
)
ax.set_yticks(range(0, num_layers, 4))
plt.colorbar(im, ax=ax, shrink=0.8, label="Probability")

# Right: Top-1 confidence at each position
ax = axes[1]
im2 = ax.imshow(top1_prob_matrix, aspect="auto", cmap="viridis", interpolation="nearest", vmin=0, vmax=1)
ax.set_xlabel("Token Position")
ax.set_ylabel("Layer")
ax.set_title("Top-1 prediction confidence at each layer")
ax.set_xticks(range(seq_len))
ax.set_xticklabels(tokens, rotation=45, ha="right", fontsize=8)
ax.set_yticks(range(0, num_layers, 4))
plt.colorbar(im2, ax=ax, shrink=0.8, label="Probability")

# Annotate with predicted tokens (last position only, every 2 layers)
last_pos = seq_len - 1
for li in range(0, num_layers, 2):
    txt = top1_text[li, last_pos]
    if len(txt) > 8:
        txt = txt[:8]
    ax.text(last_pos, li, txt, ha="center", va="center", fontsize=6,
            color="white" if top1_prob_matrix[li, last_pos] < 0.5 else "black",
            fontweight="bold")

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "g4_03_logit_lens.png"), dpi=150, bbox_inches="tight")
plt.show()
print("Saved: g4_03_logit_lens.png")
```

## Step 4: Prediction Evolution for the Last Token

Let's zoom in on how the model's top prediction for the **next token** (after "is") evolves across all 42 layers. We track:
1. The top-5 predictions at each layer
2. The probability of the final answer across layers
3. The entropy (uncertainty) at each layer

```python
# Detailed view: top-5 predictions at each layer for the last position
last_pos = seq_len - 1
print("How the prediction for '{}' -> ??? evolves:".format(tokens[last_pos].strip()))
print("=" * 80)

layer_top5 = []
with torch.no_grad():
    for li in range(num_layers):
        hidden = layer_hidden_states[li]
        if hidden.dim() == 2:
            hidden = hidden.unsqueeze(0)
        normed = final_norm(hidden)
        logits = lm_head(normed)
        if logits.dim() == 2:
            logits = logits.unsqueeze(0)
        probs = F.softmax(logits[0, last_pos].float(), dim=-1)
        top_probs, top_ids = probs.topk(5)
        top5 = [(processor.decode([tid.item()]).strip(), p.item()) for tid, p in zip(top_ids, top_probs)]
        layer_top5.append(top5)
        
        top5_str = " | ".join(["{} ({:.3f})".format(t, p) for t, p in top5])
        print("L{:>2}: {}".format(li, top5_str))
```

```python
# Track probability of the final answer across layers
# Get the final model's top prediction
final_top_token = layer_top5[-1][0][0]
# Encode the final top token to get its ID
final_top_id = processor.tokenizer.encode(final_top_token, add_special_tokens=False)[0]

prob_of_answer = []
entropy_per_layer = []

with torch.no_grad():
    for li in range(num_layers):
        hidden = layer_hidden_states[li]
        if hidden.dim() == 2:
            hidden = hidden.unsqueeze(0)
        normed = final_norm(hidden)
        logits = lm_head(normed)
        if logits.dim() == 2:
            logits = logits.unsqueeze(0)
        probs = F.softmax(logits[0, last_pos].float(), dim=-1)
        prob_of_answer.append(probs[final_top_id].cpu().item())
        # Entropy (measures uncertainty)
        ent = -(probs * torch.log(probs + 1e-10)).sum().item()
        entropy_per_layer.append(ent)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle("Prediction Crystallization: '{}'".format(PROMPT), fontsize=14, fontweight="bold")

# Left: probability of final answer
ax = axes[0]
ax.plot(range(num_layers), prob_of_answer, "o-", color="#4361ee", linewidth=2, markersize=4)
ax.fill_between(range(num_layers), prob_of_answer, alpha=0.2, color="#4361ee")
ax.set_xlabel("Layer")
ax.set_ylabel("P('{}')".format(final_top_token))
ax.set_title("Probability of '{}' at each layer".format(final_top_token))
ax.set_ylim(0, 1)
# Mark where it crosses 50%
for i, p in enumerate(prob_of_answer):
    if p > 0.5:
        ax.axvline(x=i, color="#f72585", linestyle="--", alpha=0.5)
        ax.text(i + 0.5, 0.9, "Commits at L{}".format(i), fontsize=10, color="#f72585")
        break

# Right: entropy (uncertainty)
ax = axes[1]
ax.plot(range(num_layers), entropy_per_layer, "s-", color="#f72585", linewidth=2, markersize=4)
ax.fill_between(range(num_layers), entropy_per_layer, alpha=0.2, color="#f72585")
ax.set_xlabel("Layer")
ax.set_ylabel("Entropy (nats)")
ax.set_title("Prediction Uncertainty (entropy drops = model gets confident)")

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "g4_04_logit_lens_evolution.png"), dpi=150, bbox_inches="tight")
plt.show()
print("Saved: g4_04_logit_lens_evolution.png")
```

## Step 5: Multi-Prompt Logit Lens

Let's run the logit lens on several factual prompts to see if the model always crystallizes at the same depth, or if different facts emerge at different layers.

```python
prompts = [
    "The capital of France is",
    "The largest planet in our solar system is",
    "Water boils at",
    "The CEO of Tesla is",
    "Python was created by",
    "The speed of light is approximately",
]

all_prob_curves = {}

for prompt in prompts:
    inp = processor(text=prompt, return_tensors="pt").to(model.device)
    
    # Capture hidden states
    states = {}
    handles = []
    for i in range(num_layers):
        def make_h(idx):
            def h(module, input, output):
                hidden = output[0] if isinstance(output, tuple) else output
                states[idx] = hidden.detach()
            return h
        handles.append(model.model.language_model.layers[i].register_forward_hook(make_h(i)))
    
    with torch.no_grad():
        out = model(**inp)
    
    for h in handles:
        h.remove()
    
    # Get final prediction
    final_logits = out.logits[0, -1]
    final_top_id = final_logits.argmax().item()
    final_token = processor.decode([final_top_id]).strip()
    
    # Track probability of that token across layers
    probs_curve = []
    with torch.no_grad():
        for li in range(num_layers):
            hidden = states[li]
            if hidden.dim() == 2:
                hidden = hidden.unsqueeze(0)
            normed = final_norm(hidden)
            logits = lm_head(normed)
            if logits.dim() == 2:
                logits = logits.unsqueeze(0)
            p = F.softmax(logits[0, -1].float(), dim=-1)
            probs_curve.append(p[final_top_id].cpu().item())
    
    label = "{} -> {}".format(prompt, final_token)
    all_prob_curves[label] = probs_curve
    print("Done: {}".format(label))

print("\nAll {} prompts processed.".format(len(prompts)))
```

```python
# Plot all curves
fig, ax = plt.subplots(figsize=(14, 7))
colors = plt.cm.tab10(np.linspace(0, 1, len(all_prob_curves)))

for (label, curve), color in zip(all_prob_curves.items(), colors):
    short_label = label if len(label) < 50 else label[:47] + "..."
    ax.plot(range(num_layers), curve, "o-", color=color, linewidth=2, markersize=4, label=short_label)

ax.axhline(y=0.5, color="gray", linestyle="--", alpha=0.5, label="50% threshold")
ax.set_xlabel("Layer")
ax.set_ylabel("Probability of Final Answer")
ax.set_title("Logit Lens: When Does Gemma 4 'Know' the Answer?")
ax.legend(fontsize=8, loc="upper left")
ax.set_ylim(0, 1)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "g4_05_logit_lens_multi.png"), dpi=150, bbox_inches="tight")
plt.show()
print("Saved: g4_05_logit_lens_multi.png")
```

## Key Takeaways

1. **Early layers are noisy** -- the logit lens shows near-random predictions in layers 0-8. With 42 layers, Gemma 4 has a longer "warm-up" phase where representations are still forming.

2. **Mid layers begin to converge** -- by layers 15-25, the model often has the right answer in its top-5. This is the zone where factual knowledge starts to crystallize.

3. **Final layers sharpen** -- the last 10-15 layers primarily increase confidence rather than changing the prediction. Gemma 4's deeper architecture means more refinement steps.

4. **Different facts crystallize at different depths** -- simple facts (capitals) emerge earlier than complex ones (speed of light). The 42-layer depth gives us finer resolution to observe this gradient.

5. **Entropy drops monotonically** -- the model gets steadily more confident as depth increases, with the sharpest drops often occurring in the middle third of the network.
